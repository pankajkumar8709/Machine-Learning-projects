# Random Forest Classifier (Revision Notes)

# What is Random Forest?

Random Forest is an **ensemble learning algorithm** that combines multiple **Decision Trees** to improve classification performance.

**Main Idea:**
- Train many Decision Trees.
- Each tree learns from a different random subset of the data.
- Final prediction is made by **Majority Voting**.

**Goal:** Reduce **overfitting**, reduce **variance**, and improve **generalization**.

---

# How Random Forest Works

### Step 1: Bootstrap Sampling
- Create multiple random datasets using **sampling with replacement**.

### Step 2: Train Decision Trees
- Train one Decision Tree on each bootstrap sample.

### Step 3: Random Feature Selection
- At every split, each tree considers only a **random subset of features** instead of all features.

### Step 4: Majority Voting
- Every tree predicts a class.
- Final prediction = **Class with the highest number of votes**.

# Best Performance On

Random Forest works best on:

- Structured (Tabular) datasets ✅
- Medium to Large datasets
- Mixed Numerical + Categorical features
- Non-linear relationships
- High-dimensional datasets
- Datasets with noise
- Imbalanced datasets (with `class_weight`)

### Examples

- Loan Prediction
- Credit Card Fraud Detection
- Heart Disease Prediction
- Diabetes Prediction
- Customer Churn Prediction
- Employee Attrition
- Cancer Classification
- Agriculture Prediction
- Spam Detection

---

# Not Recommended For

- Image Classification (CNN performs better)
- Text Classification (Transformers/BERT perform better)
- Very Small datasets
- Time Series forecasting (without feature engineering)

---

# Advantages

- High Accuracy
- Reduces Overfitting
- Handles Missing Values
- Works with Large Datasets
- Handles Numerical & Categorical Data
- Robust to Noise
- Provides Feature Importance
- Less Hyperparameter Tuning than a single Decision Tree

---

# Limitations

- Slower Training
- Higher Memory Usage
- Less Interpretable
- Large models can be computationally expensive

---

# Important Hyperparameters

| Hyperparameter | Description | Good Starting Values |
|---------------|-------------|----------------------|
| `n_estimators` | Number of trees | **100–300** |
| `criterion` | Split quality | **gini**, entropy |
| `max_depth` | Maximum tree depth | **10–30**, None |
| `min_samples_split` | Minimum samples to split | **2, 5, 10** |
| `min_samples_leaf` | Minimum samples in leaf | **1, 2, 4** |
| `max_features` | Features considered per split | **sqrt (recommended)**, log2 |
| `bootstrap` | Sampling with replacement | **True** |
| `class_weight` | Handles imbalance | balanced |
| `random_state` | Reproducibility | 42 |
| `n_jobs` | CPU cores | -1 |

---

# Hyperparameter Tuning Priority

Tune these first:

1. `n_estimators`
2. `max_depth`
3. `max_features`
4. `min_samples_split`
5. `min_samples_leaf`

Usually keep:

- `bootstrap=True`
- `random_state=42`
- `n_jobs=-1`

# Complexity

Training:

```
O(n_estimators × n_samples × log(n_samples))
```

Prediction:

```
O(n_estimators × tree_depth)
```

# Interview Points

- Random Forest = Ensemble of Decision Trees
- Uses **Bootstrap Sampling**
- Uses **Random Feature Selection**
- Final prediction = **Majority Voting**
- Reduces **Variance**
- Less prone to overfitting than a Decision Tree
- Works best for **Tabular Data**
- Can estimate **Feature Importance**
- Parallelizable (`n_jobs=-1`)
- More trees generally improve stability but increase training time

---

# Difference: Decision Tree vs Random Forest

| Decision Tree | Random Forest |
|--------------|---------------|
| Single Tree | Multiple Trees |
| High Variance | Low Variance |
| Easily Overfits | Less Overfitting |
| Faster | Slightly Slower |
| Lower Accuracy | Higher Accuracy |
| Easy to Interpret | Harder to Interpret |

---

# Formula

For Classification:

```
Final Prediction = Majority Vote of all Trees
```

````python
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import GridSearchCV

# Model
rf = RandomForestClassifier(random_state=42)

# Parameter Grid
param_grid = {
    'n_estimators': [200, 300],
    'criterion': ['gini', 'entropy'],
    'max_depth': [10, 20],
    'min_samples_split': [2],
    'min_samples_leaf': [1, 4],
    'max_features': ['sqrt']
}

# Grid Search
grid = GridSearchCV(
    estimator=rf,
    param_grid=param_grid,
    cv=5,
    scoring='accuracy',
    n_jobs=-1,
    verbose=2
)

grid.fit(x_train_scaled, y_train)

print("Best Parameters:", grid.best_params_)
print("Best CV Score:", grid.best_score_)

best_rf = grid.best_estimator_

y_pred = best_rf.predict(x_test_scaled)
````

````python
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import RandomizedSearchCV

# Model
rf = RandomForestClassifier(random_state=42)

# Parameter Distribution
param_dist = {
    'n_estimators': [100, 200, 300, 500],
    'criterion': ['gini', 'entropy'],
    'max_depth': [10, 20, 30, 50, 70, None],
    'min_samples_split': [2, 5, 10],
    'min_samples_leaf': [1, 2, 4],
    'max_features': ['sqrt', 'log2']
}

# Randomized Search
random = RandomizedSearchCV(
    estimator=rf,
    param_distributions=param_dist,
    n_iter=20,          # Number of random combinations
    cv=5,
    scoring='accuracy',
    random_state=42,
    n_jobs=-1,
    verbose=2
)

random.fit(x_train_scaled, y_train)

print("Best Parameters:", random.best_params_)
print("Best CV Score:", random.best_score_)

best_rf = random.best_estimator_

y_pred = best_rf.predict(x_test_scaled)
````
