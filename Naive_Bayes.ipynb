```python
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
from sklearn.naive_bayes import GaussianNB

# Load dataset
iris = load_iris()

# Take only two features
X = iris.data[:, :2]   # Sepal Length and Sepal Width
y = iris.target

# Train model
model = GaussianNB()
model.fit(X, y)

# Scatter plot
plt.figure(figsize=(8,6))
plt.scatter(X[:, 0], X[:, 1], c=y, cmap='viridis', edgecolors='k')

plt.xlabel("Sepal Length")
plt.ylabel("Sepal Width")
plt.title("Scatter Plot of Iris Dataset")

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Create meshgrid
x_min, x_max = X[:,0].min()-1, X[:,0].max()+1
y_min, y_max = X[:,1].min()-1, X[:,1].max()+1

xx, yy = np.meshgrid(
    np.arange(x_min, x_max, 0.02),
    np.arange(y_min, y_max, 0.02)
)

# Predict every point on grid
Z = model.predict(np.c_[xx.ravel(), yy.ravel()])
Z = Z.reshape(xx.shape)

# Plot decision regions
plt.figure(figsize=(8,6))
plt.contourf(xx, yy, Z, alpha=0.3, cmap='viridis')

# Plot actual points
plt.scatter(X[:,0], X[:,1], c=y, edgecolors='k', cmap='viridis')

plt.xlabel("Sepal Length")
plt.ylabel("Sepal Width")
plt.title("Gaussian Naive Bayes Decision Boundary")

plt.show()
```

Error: ValueError: 2
Description

This error occurs when you try to assign a 2-dimensional (2D) array to a 1-dimensional (1D) pandas Series.

In your case:

X_train['Embarked'] = imp2.fit_transform(X_train[['Embarked']])
What happens?

SimpleImputer.fit_transform() always returns a 2D NumPy array, even if only one column is selected.

imp2.fit_transform(X_train[['Embarked']])

Output shape:

(n_samples, 1)

But,

X_train['Embarked']

expects a 1D Series with shape:

(n_samples,)

Since pandas cannot assign a 2D array to a 1D column, it raises:

ValueError: 2

Here, 2 indicates that the input has 2 dimensions (ndim = 2).

Visual Explanation

 ['C']

 ['S']]

(4, 1)

X_train['Embarked']

Expected shape:

(4,)

ValueError: 2
How to Fix It

X_train['Embarked'] = imp2.fit_transform(X_train[['Embarked']]).ravel()
