# Bagging Regressor (Revision Notes)

## What is Bagging Regressor?

Bagging (Bootstrap Aggregating) is an **ensemble learning algorithm** used for regression. It trains multiple base models on different random subsets of the training data and **averages their predictions** to produce the final output.

**Main Goal:** Reduce **variance** and improve model stability.

---

## How It Works

1. Create multiple **bootstrap samples** (sampling **with replacement**).
2. Train one base estimator on each sample.
3. Each estimator makes a prediction.
4. Final prediction = **Average of all predictions**.

---

## When is it Beneficial?

Use Bagging Regressor when:

- Base model has **high variance** (e.g., Decision Tree).
- Model is **overfitting**.
- Dataset contains **noise**.
- Need better **generalization** and stable predictions.

**Best Base Estimator:** `DecisionTreeRegressor`

---

## Advantages

- Reduces overfitting
- Reduces variance
- Improves prediction accuracy
- More stable than a single model
- Supports parallel training

---

## Limitations

- Does **not reduce bias**
- Higher training time
- Uses more memory
- Less interpretable

---

# Important Hyperparameters

| Hyperparameter | Description |
|---------------|-------------|
| `estimator` | Base model used (e.g., `DecisionTreeRegressor()`) |
| `n_estimators` | Number of base estimators |
| `max_samples` | Number/fraction of samples for each estimator |
| `max_features` | Number/fraction of features used by each estimator |
| `bootstrap` | Sample with replacement (`True` by default) |
| `bootstrap_features` | Sample features with replacement |
| `oob_score` | Uses Out-of-Bag samples for validation |
| `random_state` | Controls randomness |
| `n_jobs` | Number of CPU cores (`-1` = all cores) |
| `verbose` | Displays training progress |

---

## Interview Points

- **Bagging = Bootstrap Aggregating**
- **Main purpose:** Reduce **Variance**
- **Prediction:** Average of all estimators
- **Bootstrap Sampling:** Sampling **with replacement**
- **Best base model:** Decision Tree Regressor
- **OOB (Out-of-Bag):** Internal validation using unused samples
- **Bagging reduces Variance, NOT Bias**

---

## Formula

\[
\hat{y}=\frac{1}{n}\sum_{i=1}^{n}\hat{y_i}
\]

Final prediction = Average of predictions from all estimators.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import kagglehub
```

```python

path=kagglehub.dataset_download("shivam2503/diamonds")
path
```

```python

import os
os.listdir(path)
```

```python

df=pd.read_csv(path+"/diamonds.csv")
```

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder,StandardScaler



le=LabelEncoder()
cat_cols = df.select_dtypes(include='object').columns

for col in cat_cols:
    df[col] = le.fit_transform(df[col])
```

```python
df
```

```python
df.drop("Unnamed: 0",axis=1,inplace=True)
```

```python
x_train,x_test,y_train,y_test=train_test_split(df.drop("price",axis=1),df["price"],test_size=0.2,random_state=42)
```

```python

scale=StandardScaler()
x_train=scale.fit_transform(x_train)
x_test=scale.transform(x_test)
```

```python
from sklearn.ensemble import BaggingRegressor
from sklearn.tree import DecisionTreeRegressor
from sklearn.metrics import mean_squared_error,r2_score
from sklearn.model_selection import GridSearchCV
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.neighbors import KNeighborsRegressor
```

```python
br=BaggingRegressor(estimator=DecisionTreeRegressor(),n_estimators=100,random_state=42)
br.fit(x_train,y_train)
```

```python
y_pred=br.predict(x_test)
r2_score(y_test,y_pred)
```

```python
br2=BaggingRegressor(estimator=DecisionTreeRegressor(),n_estimators=500,random_state=42)
br2.fit(x_train,y_train)
```

```python
y_pred2=br2.predict(x_test)
r2_score(y_test,y_pred2)
```

```python
br3=BaggingRegressor(estimator=LinearRegression(),n_estimators=200,random_state=42)
br3.fit(x_train,y_train)
```

```python
y_pred3=br3.predict(x_test)
r2_score(y_test,y_pred3)
#
```

```python
#KNeighborsRegressor
br4=BaggingRegressor(estimator=KNeighborsRegressor(),n_estimators=200,random_state=42)
br4.fit(x_train,y_train)
```

```python
y_pred4=br4.predict(x_test)
r2_score(y_test,y_pred4)
#
```

```python
from sklearn.svm import SVR
```

```python
br5=BaggingRegressor(estimator=SVR(),n_estimators=200,random_state=42)
br5.fit(x_train,y_train)
```

```python
y_pred5=br5.predict(x_test)
r2_score(y_test,y_pred5)
```

```python
y_pred6=dt.predict(x_test)
r2_score(y_test,y_pred6)
```
